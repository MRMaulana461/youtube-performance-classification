```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Data dari classification report
metrics_df = pd.DataFrame({
    'Kelas': ['Rendah', 'Sedang', 'Tinggi'],
    'Precision': [0.95, 0.90, 0.60],
    'Recall': [0.91, 0.93, 0.73],
    'F1-Score': [0.93, 0.91, 0.66]
})

# Posisi bar
x = np.arange(len(metrics_df['Kelas']))
width = 0.25

plt.figure(figsize=(10, 6))

plt.bar(x - width, metrics_df['Precision'], width, label='Precision')
plt.bar(x, metrics_df['Recall'], width, label='Recall')
plt.bar(x + width, metrics_df['F1-Score'], width, label='F1-Score')

plt.xlabel('Kelas')
plt.ylabel('Nilai')
plt.title('Perbandingan Precision, Recall, dan F1-Score per Kelas')
plt.xticks(x, metrics_df['Kelas'])
plt.ylim(0, 1.05)

# Menampilkan nilai di atas batang
for i in range(len(metrics_df)):
    plt.text(x[i]-width, metrics_df['Precision'][i]+0.01,
             f"{metrics_df['Precision'][i]:.2f}",
             ha='center')

    plt.text(x[i], metrics_df['Recall'][i]+0.01,
             f"{metrics_df['Recall'][i]:.2f}",
             ha='center')

    plt.text(x[i]+width, metrics_df['F1-Score'][i]+0.01,
             f"{metrics_df['F1-Score'][i]:.2f}",
             ha='center')

plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Data Generalization Test
metrics_df = pd.DataFrame({
    'Kelas': ['Rendah', 'Sedang', 'Tinggi'],
    'Precision': [1.00, 0.90, 0.00],
    'Recall': [0.76, 0.96, 0.00],
    'F1-Score': [0.86, 0.93, 0.00]
})

x = np.arange(len(metrics_df))
width = 0.25

plt.figure(figsize=(10,6))

bars1 = plt.bar(x - width, metrics_df['Precision'], width, label='Precision')
bars2 = plt.bar(x, metrics_df['Recall'], width, label='Recall')
bars3 = plt.bar(x + width, metrics_df['F1-Score'], width, label='F1-Score')

plt.title('Classification Report - Generalization Test')
plt.xlabel('Kelas')
plt.ylabel('Skor')
plt.xticks(x, metrics_df['Kelas'])
plt.ylim(0, 1.1)

for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width()/2,
            height + 0.02,
            f'{height:.2f}',
            ha='center'
        )

plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
